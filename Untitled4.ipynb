```python
from sklearn.preprocessing import StandardScaler

from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

from sklearn.model_selection import (
    train_test_split,
    StratifiedKFold,
    RandomizedSearchCV
)

from sklearn.pipeline import Pipeline

from sklearn.linear_model import LogisticRegression

from sklearn.ensemble import (
    RandomForestClassifier,
    ExtraTreesClassifier
)
!pip install xgboost
from xgboost import XGBClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report,
    ConfusionMatrixDisplay,
    roc_curve
)
!pip install shap
import shap
import joblib

import warnings
warnings.filterwarnings("ignore")

sns.set_theme(style="whitegrid")

print("All libraries imported successfully!")
```

```python
rfm_scaler = StandardScaler()

rfm_scaled = rfm_scaler.fit_transform(
    rfm_log
)
```

```python
inertia = []
silhouette_scores = []

k_values = range(2, 9)

for k in k_values:

    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=20
    )

    labels = kmeans.fit_predict(
        rfm_scaled
    )

    inertia.append(
        kmeans.inertia_
    )

    silhouette_scores.append(
        silhouette_score(
            rfm_scaled,
            labels
        )
    )
```

```python
## Elbow Method 
```

```python
plt.figure(figsize=(8, 5))

plt.plot(
    list(k_values),
    inertia,
    marker="o"
)

plt.title("Elbow Method")
plt.xlabel("Number of Clusters")
plt.ylabel("Inertia")

plt.show()
```

## Silhoutte Score

```python
plt.figure(figsize=(8, 5))

plt.plot(
    list(k_values),
    silhouette_scores,
    marker="o"
)

plt.title("Silhouette Score")
plt.xlabel("Number of Clusters")
plt.ylabel("Silhouette Score")

plt.show()
```

```python
cutoff_date = pd.Timestamp(
    "2011-10-01"
)

train_df = df[
    df["InvoiceDate"] < cutoff_date
].copy()

future_df = df[
    df["InvoiceDate"] >= cutoff_date
].copy()

print(
    "Training transactions:",
    len(train_df)
)

print(
    "Future transactions:",
    len(future_df)
)
```

```python
train_features = (
    train_df
    .groupby("CustomerID")
    .agg(
        Recency=(
            "InvoiceDate",
            lambda x:
            (
                cutoff_date -
                x.max()
            ).days
        ),

        Frequency=(
            "InvoiceNo",
            "nunique"
        ),

        Monetary=(
            "Revenue",
            "sum"
        ),

        Total_Quantity=(
            "Quantity",
            "sum"
        ),

        Unique_Products=(
            "StockCode",
            "nunique"
        ),

        First_Purchase=(
            "InvoiceDate",
            "min"
        ),

        Last_Purchase=(
            "InvoiceDate",
            "max"
        )
    )
    .reset_index()
)
```

```python
train_features["Average_Order_Value"] = (
    train_features["Monetary"] /
    train_features["Frequency"]
)

train_features["Customer_Lifetime_Days"] = (
    train_features["Last_Purchase"] -
    train_features["First_Purchase"]
).dt.days

train_features["Purchase_Frequency"] = (
    train_features["Frequency"] /
    (
        train_features["Customer_Lifetime_Days"]
        + 1
    )
)
```

```python
future_customers = set(
    future_df["CustomerID"].unique()
)

train_features["Repeat_Purchase"] = (
    train_features["CustomerID"]
    .isin(future_customers)
    .astype(int)
)
```

```python
target_counts = (
    train_features["Repeat_Purchase"]
    .value_counts()
)

print(target_counts)

print("\nPercentage:")
print(
    train_features[
        "Repeat_Purchase"
    ].value_counts(
        normalize=True
    ) * 100
)
```

## Selecting ML features

```python
features = [
    "Recency",
    "Frequency",
    "Monetary",
    "Total_Quantity",
    "Unique_Products",
    "Average_Order_Value",
    "Customer_Lifetime_Days",
    "Purchase_Frequency"
]

X = train_features[
    features
].copy()

y = train_features[
    "Repeat_Purchase"
].copy()
```

```python
X = X.replace(
    [np.inf, -np.inf],
    np.nan
)

X = X.fillna(0)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training rows:", len(X_train))
print("Testing rows:", len(X_test))
```

```python
logistic_model = Pipeline([
    (
        "scaler",
        StandardScaler()
    ),

    (
        "model",
        LogisticRegression(
            max_iter=2000,
            class_weight="balanced",
            random_state=42
        )
    )
])

logistic_model.fit(
    X_train,
    y_train
)

logistic_pred = (
    logistic_model.predict(X_test)
)

logistic_prob = (
    logistic_model
    .predict_proba(X_test)[:, 1]
)
```

```python
rf_model = RandomForestClassifier(
    n_estimators=500,
    random_state=42,
    class_weight="balanced",
    n_jobs=-1
)

rf_model.fit(
    X_train,
    y_train
)

rf_pred = rf_model.predict(
    X_test
)

rf_prob = (
    rf_model
    .predict_proba(X_test)[:, 1]
)
```

```python
extra_model = ExtraTreesClassifier(
    n_estimators=500,
    random_state=42,
    class_weight="balanced",
    n_jobs=-1
)

extra_model.fit(
    X_train,
    y_train
)

extra_pred = extra_model.predict(
    X_test
)

extra_prob = (
    extra_model
    .predict_proba(X_test)[:, 1]
)
```

```python
xgb_model = XGBClassifier(
    n_estimators=500,
    learning_rate=0.05,
    max_depth=5,
    subsample=0.8,
    colsample_bytree=0.8,
    eval_metric="logloss",
    random_state=42,
    n_jobs=-1
)

xgb_model.fit(
    X_train,
    y_train
)

xgb_pred = xgb_model.predict(
    X_test
)

xgb_prob = (
    xgb_model
    .predict_proba(X_test)[:, 1]
)
```

```python
def evaluate_model(
    model_name,
    y_true,
    y_pred,
    y_probability
):

    return {
        "Model": model_name,

        "Accuracy": accuracy_score(
            y_true,
            y_pred
        ),

        "Precision": precision_score(
            y_true,
            y_pred,
            zero_division=0
        ),

        "Recall": recall_score(
            y_true,
            y_pred,
            zero_division=0
        ),

        "F1 Score": f1_score(
            y_true,
            y_pred,
            zero_division=0
        ),

        "ROC-AUC": roc_auc_score(
            y_true,
            y_probability
        )
    }
```

```python
results = []

results.append(
    evaluate_model(
        "Logistic Regression",
        y_test,
        logistic_pred,
        logistic_prob
    )
)

results.append(
    evaluate_model(
        "Random Forest",
        y_test,
        rf_pred,
        rf_prob
    )
)

results.append(
    evaluate_model(
        "Extra Trees",
        y_test,
        extra_pred,
        extra_prob
    )
)

results.append(
    evaluate_model(
        "XGBoost",
        y_test,
        xgb_pred,
        xgb_prob
    )
)

results_df = pd.DataFrame(
    results
)

results_df.sort_values(
    "F1 Score",
    ascending=False
)
```

```python
plt.figure(figsize=(12, 6))

results_melted = results_df.melt(
    id_vars="Model",
    value_vars=[
        "Accuracy",
        "Precision",
        "Recall",
        "F1 Score",
        "ROC-AUC"
    ],
    var_name="Metric",
    value_name="Score"
)

sns.barplot(
    data=results_melted,
    x="Model",
    y="Score",
    hue="Metric"
)

plt.title("Machine Learning Model Comparison")

plt.ylim(0, 1)

plt.xticks(rotation=20)

plt.show()
```

```python
def risk_category(probability):

    if probability >= 0.75:
        return "High Probability"

    elif probability >= 0.50:
        return "Medium Probability"

    else:
        return "Low Probability"


train_features["Purchase_Risk"] = (
    train_features[
        "Purchase_Probability"
    ].apply(risk_category)
)
```

```python
def predict_customer(
    recency,
    frequency,
    monetary,
    total_quantity,
    unique_products,
    average_order_value,
    customer_lifetime_days,
    purchase_frequency
):

    customer = pd.DataFrame([{
        "Recency": recency,
        "Frequency": frequency,
        "Monetary": monetary,
        "Total_Quantity": total_quantity,
        "Unique_Products": unique_products,
        "Average_Order_Value": average_order_value,
        "Customer_Lifetime_Days":
            customer_lifetime_days,
        "Purchase_Frequency":
            purchase_frequency
    }])

    probability = (
        tuned_model
        .predict_proba(customer)[0][1]
    )

    prediction = (
        "Likely to Purchase"
        if probability >= 0.50
        else "Unlikely to Purchase"
    )

    return probability, prediction
```

```python
probability, prediction = predict_customer(
    recency=15,
    frequency=20,
    monetary=5000,
    total_quantity=100,
    unique_products=35,
    average_order_value=250,
    customer_lifetime_days=250,
    purchase_frequency=20/251
)

print(
    "Purchase Probability:",
    round(probability * 100, 2),
    "%"
)

print(
    "Final Prediction:",
    prediction
)
```

```python
#3. Recency vs Monetary by Customer Segment
plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=customer_data,
    x="Recency",
    y="Monetary",
    hue="Segment",
    s=70
)

plt.title("Recency vs Monetary by Customer Segment")
plt.xlabel("Days Since Last Purchase")
plt.ylabel("Total Spending")

plt.show()
```

```python
# 4. Customer Segment Distribution
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(8, 5))

sns.countplot(
    data=customer_data,
    x="Segment"
)

plt.title("Customer Segment Distribution")
plt.xlabel("Customer Segment")
plt.ylabel("Number of Customers")
plt.xticks(rotation=20)
plt.show()
```

```python
# 5. Average Order Value Trend Over Customer Lifetime

# Create lifetime groups
customer_data["Lifetime_Group"] = pd.cut(
    customer_data["Customer_Lifetime_Days"],
    bins=10
)

# Calculate average AOV for each lifetime group
trend_data = customer_data.groupby(
    "Lifetime_Group",
    observed=True
)["Average_Order_Value"].mean().reset_index()

plt.figure(figsize=(10, 6))

plt.plot(
    trend_data["Lifetime_Group"].astype(str),
    trend_data["Average_Order_Value"],
    marker="o"
)

plt.title("Average Order Value Trend Over Customer Lifetime")
plt.xlabel("Customer Lifetime")
plt.ylabel("Average Order Value")

plt.xticks(rotation=45)
plt.grid(True)

plt.show()
```

```python
# 5. Average Order Value Trend Over Time

customer_data["Last_Purchase"] = pd.to_datetime(
    customer_data["Last_Purchase"]
)

trend_data = (
    customer_data
    .set_index("Last_Purchase")
    .resample("M")["Average_Order_Value"]
    .mean()
    .reset_index()
)

plt.figure(figsize=(12, 6))

plt.plot(
    trend_data["Last_Purchase"],
    trend_data["Average_Order_Value"],
    marker="o"
)

plt.title("Average Order Value Trend Over Time")
plt.xlabel("Month")
plt.ylabel("Average Order Value")

plt.xticks(rotation=45)
plt.grid(True)

plt.show()
```
